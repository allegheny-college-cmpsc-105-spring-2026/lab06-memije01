```python
# data_analytics.csv

# : update the column names to match the names you have in your csv file for data_analytics

# create a pandas dataframe
dfa = pd.read_csv(
    '/content/data_analytics.csv',
    sep=',',
    names = ['Views', 'Active Users', 'Views per active user', 'Average engagement time per active user', 'Event count', 'Key events', 'Total revenue'],
    skiprows=10
)

# Display the first few rows of the DataFrame
print(dfa.head())
```

```python
# data_survey.csv

# TODO: update the column names to match the names you have in your csv file for data_survey

# create a pandas dataframe
dfs = pd.read_csv(
    '/content/data_survey.csv',
    sep=',',
    names = ['Total_score', 'Do you play any sports', 'Do you watch sports?'],
    skiprows=1
)

# Display the first few rows of the DataFrame
print(dfs.head())
```

```python
# : how many rows and columns are in the dataframe dfa? (3,7)
# : how many rows and columns are in the dataframe dfs? (15,3)
```

```python
# Analytics

# : Do one or more explorations with dfa
# no interaction is required

# make a blank canvas

# make a title starting with 'Figure 1:' Fill in a relevant title

# plot
# ylabel
# xlabel
# ylim
# etc.

# reveal everything

import matplotlib.pyplot as plt

# make a blank canvas
plt.figure(figsize=(10, 6))

# plot
plt.bar(dfa.index, dfa['Views'])

# title and labels
plt.title('Figure 1: Views per URL Path')
plt.ylabel('Number of Views')
plt.xlabel('URL Path')

# rotate x-axis labels for readability
plt.xticks(rotation=45, ha='right')

# reveal everything
plt.tight_layout()
plt.show()
```

```python
# Survey

# : Do one or more explorations with dfs
# no interaction is required

import matplotlib.pyplot as plt

# make a blank canvas
plt.figure(figsize=(8, 5))

# Get value counts for the 'Total_score' column (which contains 'Yes'/'No' responses)
response_counts = dfs['Total_score'].value_counts()


# make a title starting with 'Figure 2:' Fill in a relevant title
plt.title('Figure 2: Distribution of Responses (from Total_score column)')
plt.xlabel('Response')
plt.ylabel('Number of Respondents')
plt.bar(response_counts.index, response_counts.values, color=['skyblue', 'lightcoral'])


# reveal everything
plt.tight_layout()
plt.show()
```
